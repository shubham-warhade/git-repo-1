```python
import pandas as pd

data1 = pd.read_csv("Records Matching Task.xlsx - data1.csv")
data2 = pd.read_csv("Records Matching Task.xlsx - data2.csv")
```

```python
import pandas as pd
```

```python
data1["Key"] = data1["Order ID"].astype(str) + "_" + data1["Product ID"].astype(str)
data2["Key"] = data2["Order ID"].astype(str) + "_" + data2["Product ID"].astype(str)
```

```python
set1 = set(data1["Key"])
set2 = set(data2["Key"])
```

```python
missing_in_data2 = data1[data1["Key"].isin(set1 - set2)]

print("Records present in Data1 but missing in Data2:")
print(len(missing_in_data2))
```

```python
missing_in_data1 = data2[data2["Key"].isin(set2 - set1)]

print("Records present in Data2 but missing in Data1:")
print(len(missing_in_data1))
```

```python
qty_sum = missing_in_data1["Qty"].sum()

print("Sum of Qty:", qty_sum)
```

```python
total_unique = len(set1.union(set2))

print("Total Unique Records:", total_unique)
```

```python
missing_in_data2.to_csv("data1_missing_in_data2.csv", index=False)
missing_in_data1.to_csv("data2_missing_in_data1.csv", index=False)
```

```python
memory=df.memory_usage(deep=True).sum()
print(memory)
```

```python
for col in df.select_dtypes(include=['int64']).columns:
  df[col]=pd.to_numeric(df[col],downcast='integer')
```

```python
for col in df.select_dtypes(include=['float64']).columns:
    df[col] = pd.to_numeric(df[col], downcast='float')

for col in df.select_dtypes(include=['object']).columns:
    if df[col].nunique() / len(df) < 0.5:
        df[col] = df[col].astype('category')
```

```python
memory_before = df.memory_usage(deep=True).sum() / 1024**2
print(f"Memory Usage Before: {memory_before:.2f} MB")
```

```python
memory_after = df.memory_usage(deep=True).sum() / 1024**2

print(f"Memory Usage After : {memory_after:.2f} MB")
print(f"Memory Saved       : {memory_before-memory_after:.2f} MB")
```
